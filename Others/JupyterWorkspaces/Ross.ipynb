```python
parts = [2,2,2]
G = graphs.CompleteMultipartiteGraph(parts)
# G = graphs.CompleteGraph(4)
G.show()
M = Matroid(G)
L = M.lattice_of_flats()
L.show(figsize=[100,100])

Sub_Decomp = []
s = M.hyperplanes()

# Create the list of subdirect decompositions
for i in s:
    decomp = L.sublattice(L.order_ideal([i])).subdirect_decomposition()
    Sub_Decomp.append(decomp)

# Display each lattice in order
for i, decomp in enumerate(Sub_Decomp):
    print(f"Subdirect Decomposition {i}:")
    for j, lattice in enumerate(decomp):
        print(f"  Lattice {j}:")
        lattice.show()
```
