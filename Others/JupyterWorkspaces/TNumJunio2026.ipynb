# Factorización de Ideales en Dominios de Dedekind

Consideremos el cuerpo cuadrático $L = \mathbb{Q}(\sqrt{5})$ y su anillo de enteros $\mathcal{O}_L$. 
Como $5 \equiv 1 \pmod 4$, el anillo de enteros es:
$$\mathcal{O}_L = \mathbb{Z}[\theta], \quad \text{donde } \theta = \frac{1+\sqrt{5}}{2}$$

El polinomio mínimo de $\theta$ sobre $\mathbb{Q}$ es:
$$f(x) = x^2 - x - 1$$

Para estudiar cómo se factoriza un ideal generado por un primo $p$ en $\mathcal{O}_L$ (es decir, $p\mathcal{O}_L$), utilizamos el **Teorema de Kummer-Dedekind**.

---

### Teorema (Kummer-Dedekind)
Sea $L = \mathbb{Q}(\theta)$ un cuerpo de números con $\mathcal{O}_L = \mathbb{Z}[\theta]$, y sea $f(x) \in \mathbb{Z}[x]$ el polinomio mínimo de $\theta$. Sea $p \in \mathbb{Z}$ un número primo. 

Si la reducción de $f(x)$ módulo $p$ se factoriza en $(\mathbb{Z}/p\mathbb{Z})[x]$ como:
$$\bar{f}(x) \equiv \bar{g}_1(x)^{e_1} \bar{g}_2(x)^{e_2} \cdots \bar{g}_r(x)^{e_r} \pmod p$$
donde los $\bar{g}_i(x)$ son polinomios mónicos e irreducibles módulo $p$, entonces la descomposición en ideales primos del ideal $p\mathcal{O}_L$ es:
$$p\mathcal{O}_L = \mathfrak{p}_1^{e_1} \mathfrak{p}_2^{e_2} \cdots \mathfrak{p}_r^{e_r}$$
donde los ideales primos están dados por:
$$\mathfrak{p}_i = p\mathcal{O}_L + g_i(\theta)\mathcal{O}_L = \langle p, g_i(\theta) \rangle$$
y sus grados de inercia (o residuo) son $f_i = \deg(g_i)$.

## Caso $p = 11$ (Escisión / Splits)

Para factorizar el ideal $11\mathcal{O}_L$, primero factorizamos el polinomio mínimo $f(x) = x^2 - x - 1$ en $\mathbb{F}_{11}[x]$:
$$x^2 - x - 1 \equiv (x - 4)(x - 8) \pmod{11}$$

En este caso, tenemos:
* $g_1(x) = x - 4 \implies g_1(\theta) = \theta - 4 = \frac{1+\sqrt{5}}{2} - 4 = \frac{-7+\sqrt{5}}{2}$
* $g_2(x) = x - 8 \implies g_2(\theta) = \theta - 8 = \frac{1+\sqrt{5}}{2} - 8 = \frac{-15+\sqrt{5}}{2}$

Por el teorema, el ideal se descompone como el producto de dos ideales primos distintos:
$$11\mathcal{O}_L = \mathfrak{b}_1 \mathfrak{b}_2$$
donde:
* $\mathfrak{b}_1 = \langle 11, \frac{-7+\sqrt{5}}{2} \rangle$
* $\mathfrak{b}_2 = \langle 11, \frac{-15+\sqrt{5}}{2} \rangle$

A continuación, verificamos esto usando SageMath.

````sage
from sage.all import GF, NumberField, QQ, factor

# 1. Factorización del polinomio mínimo en GF(11)
R = GF(11)['x']
x = R.gen()
f = x**2 - x - 1
print("Factorización de f(x) mod 11:")
print(factor(f))

# 2. Factorización directa del ideal (11) en el cuerpo de números Q(sqrt(5))
Qx = QQ['x']
x_QQ = Qx.gen()
L = NumberField(x_QQ**2 - 5, names='w')
w = L.gen()
print("\nFactorización del ideal (11) en L:")
print(L.ideal(11).factor())
````

## Ejercicio: Factorizar $p=2$ y $p=5$

Aplicamos el mismo criterio para los primos $p=2$ y $p=5$.

### Caso $p = 2$ (Inerte / Inert)
Analizamos la reducción de $f(x) = x^2 - x - 1$ módulo $2$:
* Módulo 2, el polinomio es $x^2 + x + 1$. Evaluando en los elementos de $\mathbb{F}_2$:
  * Para $x = 0$: $0^2 + 0 + 1 = 1 \not\equiv 0 \pmod 2$.
  * Para $x = 1$: $1^2 + 1 + 1 = 3 \equiv 1 \not\equiv 0 \pmod 2$.
* Como no tiene raíces en $\mathbb{F}_2$ y es de grado 2, el polinomio $\bar{f}(x)$ es irreducible módulo 2.
* Por lo tanto, el ideal $2\mathcal{O}_L$ se mantiene primo (es decir, $2$ es **inerte** en $\mathcal{O}_L$).

---

### Caso $p = 5$ (Ramificado / Ramified)
Analizamos la reducción de $f(x) = x^2 - x - 1$ módulo $5$:
* Notemos que módulo 5, podemos completar el cuadrado:
  $$x^2 - x - 1 \equiv x^2 - 6x + 9 \equiv (x - 3)^2 \pmod 5$$
* Como el polinomio se reduce a un cuadrado perfecto, el ideal $5\mathcal{O}_L$ **ramifica**:
  $$5\mathcal{O}_L = \mathfrak{p}^2$$
  donde el ideal primo es:
  $$\mathfrak{p} = \langle 5, \theta - 3 \rangle = \langle 5, \frac{1+\sqrt{5}}{2} - 3 \rangle = \langle 5, \frac{-5+\sqrt{5}}{2} \rangle$$

Verifiquemos estos resultados en SageMath.

````sage
from sage.all import GF, NumberField, QQ, factor

# Factorización de f(x) módulo 2 y 5
R2 = GF(2)['x']
x2 = R2.gen()
f2 = x2**2 - x2 - 1

R5 = GF(5)['x']
x5 = R5.gen()
f5 = x5**2 - x5 - 1

print("Factorización mod 2:")
print(f"{f2} = {factor(f2)}")

print("\nFactorización mod 5:")
print(f"{f5} = {factor(f5)}")

# Factorización de los ideales en el cuerpo de números
Qx = QQ['x']
x_QQ = Qx.gen()
L = NumberField(x_QQ**2 - 5, names='w')
w = L.gen()

print("\nFactorización del ideal (2) en L:")
print(L.ideal(2).factor())

print("\nFactorización del ideal (5) en L:")
print(L.ideal(5).factor())
````

# Sesión 2: Números $p$-ádicos

## Introducción Práctica a los Números $p$-ádicos

Comencemos por recordar cómo escribimos habitualmente los números naturales. Utilizamos sus dígitos en base 10. Por ejemplo:
$$25 = 2 \cdot 10^1 + 5 \cdot 10^0$$

Cualquier número entero positivo se puede representar de forma única como una combinación de potencias de $10$. Sin embargo, la base 10 es solo una convención cultural; es perfectamente posible representar números en cualquier otra base entera $p > 1$ (como base 2, base 8, base 16, etc.).

Por el algoritmo de la división, todo número natural $N$ se puede expresar de forma única en base $p$ (donde $p$ es un número primo) como:
$$N = a_n p^n + a_{n-1} p^{n-1} + \dots + a_1 p^1 + a_0 p^0$$
donde los coeficientes (o dígitos) satisfacen $a_j \in \{0, 1, \dots, p-1\}$ y $a_n \neq 0$.

Representamos esto escribiendo los dígitos en orden de potencias decrecientes:
$$N = (a_n a_{n-1} \dots a_1 a_0)_p$$

### Ejemplos en base $p = 11$:

1. **Para $N = 25$**:
   $$25 = 2 \cdot 11^1 + 3 \cdot 11^0 \implies 25 = (23)_{11}$$

2. **Para $N = 345$**:
   $$345 = 2 \cdot 11^2 + 9 \cdot 11^1 + 4 \cdot 11^0 \implies 345 = (294)_{11}$$

3. **Para $N = 500$**:
   $$500 = 4 \cdot 11^2 + 1 \cdot 11^1 + 5 \cdot 11^0 \implies 500 = (415)_{11}$$

A continuación, implementamos una función en Python para calcular de manera interactiva la expansión en base $p$ de cualquier número entero.

````sage
def expansion_base_p(n, p):
    """
    Retorna la lista de dígitos de n en base p,
    de mayor a menor potencia.
    """
    if n == 0:
        return [0]
    digitos = []
    temp = n
    while temp > 0:
        digitos.append(temp % p)
        temp //= p
    return digitos[::-1]

# Pruebas de los ejemplos en base 11
print(f"25 en base 11: {expansion_base_p(25, 11)}")
print(f"345 en base 11: {expansion_base_p(345, 11)}")
print(f"500 en base 11: {expansion_base_p(500, 11)}")
````

## Operaciones Aritméticas en Base $p$

La suma y el producto de números expresados en base $p$ funcionan exactamente igual que en base 10, aplicando las reglas de llevarse unidades (arrastre) cuando un resultado parcial supera o iguala a $p$.

### Ejemplo en base $p = 11$:
Consideremos $25 = (23)_{11}$ y $37 = (34)_{11}$ (ya que $37 = 3 \cdot 11^1 + 4 \cdot 11^0$).

1. **Suma**:
   $$(23)_{11} + (34)_{11} = (57)_{11}$$
   Verificamos en base 10:
   $$(57)_{11} = 5 \cdot 11 + 7 = 62 = 25 + 37$$

2. **Producto**:
   Multiplicamos $(23)_{11} \times (34)_{11}$ dígito por dígito:
   * Multiplicando por 4: 
     * $4 \times 3 = 12 = 1 \cdot 11 + 1 \implies$ anotamos $1$, nos llevamos $1$.
     * $4 \times 2 = 8$, más $1$ que nos llevábamos $= 9$.
     * Resultado parcial: $(91)_{11}$.
   * Multiplicando por 3 (desplazado una posición):
     * $3 \times 3 = 9$.
     * $3 \times 2 = 6$.
     * Resultado parcial: $(690)_{11}$.
   * Sumamos los resultados parciales:
     $$(91)_{11} + (690)_{11} = (771)_{11}$$
     * En las unidades: $1 + 0 = 1$.
     * En las "decenas": $9 + 9 = 18 = 1 \cdot 11 + 7 \implies$ anotamos $7$, nos llevamos $1$.
     * En las "centenas": $6 + 1 \text{ (carry)} = 7$.

   Verificamos en base 10:
   $$(771)_{11} = 7 \cdot 11^2 + 7 \cdot 11 + 1 = 847 + 77 + 1 = 925 = 25 \times 37$$

---

### Ejemplos en base $p = 17$:
Para bases mayores a 10, usamos letras para representar dígitos mayores a 9. En base 17:
$$A=10,\ B=11,\ C=12,\ D=13,\ E=14,\ F=15,\ G=16$$

#### Ejemplo 1: Suma con acarreo
Consideremos $N_1 = 30$ y $N_2 = 40$:
* $30 = 1 \cdot 17 + 13 \implies (1D)_{17}$
* $40 = 2 \cdot 17 + 6 \implies (26)_{17}$

Sumamos $(1D)_{17} + (26)_{17}$:
* Suma de unidades: $D + 6 = 13 + 6 = 19 = 1 \cdot 17 + 2 \implies$ anotamos $2$, llevamos $1$.
* Suma de las siguientes posiciones: $1 + 2 + 1 \text{ (carry)} = 4$.
* Resultado: $(42)_{17}$.

Verificamos:
$$(42)_{17} = 4 \cdot 17 + 2 = 70 = 30 + 40$$

#### Ejemplo 2: Producto
Consideremos $N_1 = 20$ y $N_2 = 35$:
* $20 = 1 \cdot 17 + 3 \implies (13)_{17}$
* $35 = 2 \cdot 17 + 1 \implies (21)_{17}$

Multiplicamos $(13)_{17} \times (21)_{17}$:
* Multiplicando por 1: $(13)_{17}$.
* Multiplicando por 2 (desplazado): $2 \times 3 = 6$, $2 \times 1 = 2 \implies (260)_{17}$.
* Sumando parciales: $(13)_{17} + (260)_{17} = (273)_{17}$.

Verificamos:
$$(273)_{17} = 2 \cdot 17^2 + 7 \cdot 17 + 3 = 578 + 119 + 3 = 700 = 20 \times 35$$

````sage
def a_base_p(n, p):
    """
    Convierte un entero n a su representación de cadena en base p (hasta base 36).
    """
    if n == 0:
        return "0"
    digitos = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    resultado = ""
    temp = n
    while temp > 0:
        resultado = digitos[temp % p] + resultado
        temp //= p
    return resultado

# Validación de los ejemplos en base 11
print("--- Base 11 ---")
print(f"37 en base 11: {a_base_p(37, 11)}")
print(f"Suma 25 + 37 = 62 en base 11: {a_base_p(62, 11)}")
print(f"Producto 25 * 37 = 925 en base 11: {a_base_p(925, 11)}")

# Validación de los ejemplos en base 17
print("\n--- Base 17 ---")
print(f"30 en base 17: {a_base_p(30, 17)}")
print(f"40 en base 17: {a_base_p(40, 17)}")
print(f"Suma 30 + 40 = 70 en base 17: {a_base_p(70, 17)}")

print(f"20 en base 17: {a_base_p(20, 17)}")
print(f"35 en base 17: {a_base_p(35, 17)}")
print(f"Producto 20 * 35 = 700 en base 17: {a_base_p(700, 17)}")
````

## Expansiones Infinitas y Representación de Enteros Negativos

¿Qué ocurre si permitimos que la suma de potencias de $p$ sea infinita hacia la izquierda? Es decir, si consideramos series de la forma:
$$\sum_{k=0}^{\infty} a_k p^k = a_0 + a_1 p + a_2 p^2 + \dots$$
donde $a_k \in \{0, 1, \dots, p-1\}$.

A primera vista, esto parece no tener sentido en la aritmética usual porque la serie diverge (los términos se hacen infinitamente grandes). Sin embargo, en el mundo $p$-ádico las cosas cambian.

### El ejemplo de $-1$

Consideremos $p = 5$ y el número con infinitos dígitos a la izquierda $(\dots 444)_5 = \sum_{k=0}^{\infty} 4 \cdot 5^k$.
Si le sumamos $1$ usando la aritmética usual con acarreo:
* En las unidades: $4 + 1 = 5 \equiv 0 \pmod 5$, y llevamos $1$.
* En las siguientes posiciones: $4 + 1 \text{ (carry)} = 5 \equiv 0 \pmod 5$, y llevamos $1$.
* Este acarreo se propaga indefinidamente hacia la izquierda.

El resultado final "limita" a $0$. Por lo tanto, deducimos que:
$$(\dots 444)_5 + 1 = 0 \implies (\dots 444)_5 = -1$$

Podemos generalizar esto para cualquier primo $p$. El número formado por el dígito $p-1$ repetido infinitamente es igual a $-1$:
$$-1 = (\dots (p-1)(p-1)(p-1))_p = \sum_{k=0}^{\infty} (p-1) p^k$$

### Formalización mediante sumas parciales

Esta igualdad se puede formalizar analizando las sumas parciales $S_n$:
$$S_n = \sum_{k=0}^{n-1} (p-1) p^k$$

Usando la fórmula de la progresión geométrica:
$$S_n = (p-1) \frac{p^n - 1}{p-1} = p^n - 1$$

Notemos que para cualquier $n$, se cumple la congruencia:
$$S_n \equiv -1 \pmod{p^n}$$

En el sentido usual, $p^n \to \infty$ cuando $n \to \infty$. Pero en el sentido $p$-ádico, a medida que $n$ crece, $p^n$ es divisible por potencias cada vez más altas de $p$, lo que lo hace "cada vez más cercano a cero". Por lo tanto, el límite de las sumas parciales en la métrica $p$-ádica es:
$$\lim_{n \to \infty} S_n = \lim_{n \to \infty} (p^n - 1) = 0 - 1 = -1$$

---

## ¿En qué sentido convergen estas sumas infinitas? (Valuaciones y Métricas)

Para dar rigor a esta convergencia, definimos la **valuación $p$-ádica** y el **valor absoluto $p$-ádico**:

1. **Valuación $p$-ádica ($v_p(x)$)**: 
   Para cualquier entero $n \neq 0$, $v_p(n)$ es el exponente de la mayor potencia de $p$ que divide a $n$. Para un número racional $x = a/b$, se define como:
   $$v_p\left(\frac{a}{b}\right) = v_p(a) - v_p(b)$$
   Por convención, $v_p(0) = \infty$.

2. **Valor absoluto $p$-ádico ($|x|_p$)**:
   Se define para cualquier $x \in \mathbb{Q}$ como:
   $$|x|_p = \begin{cases} p^{-v_p(x)} & \text{si } x \neq 0 \\ 0 & \text{si } x = 0 \end{cases}$$

Bajo esta definición:
* Las potencias de $p$ son pequeñas: $|p^k|_p = p^{-k} \to 0$ cuando $k \to \infty$.
* Una serie $\sum a_k p^k$ converge en la métrica $p$-ádica si y solo si su término general tiende a cero en dicha métrica, lo cual se cumple ya que $|a_k p^k|_p \le p^{-k} \to 0$.

### Expansión de números racionales

No solo los enteros (positivos y negativos) tienen expansión $p$-ádica. También los números racionales de la forma $a/b$ con $p \nmid b$ poseen una representación.

**Ejemplo**: La suma geométrica $\sum_{k=0}^{\infty} p^k = 1 + p + p^2 + \dots$
Dado que la razón de la serie es $p$, y su valor absoluto $p$-ádico es $|p|_p = 1/p < 1$, la serie converge en $\mathbb{Q}_p$ a:
$$\frac{1}{1-p} = (\dots 111)_p$$

Por ejemplo, si $p = 5$:
$$\frac{1}{1-5} = -\frac{1}{4} = (\dots 111)_5$$

A continuación, verificamos estas representaciones y límites en SageMath usando el campo de los números $p$-ádicos $\mathbb{Q}_p$ (representado en Sage como `Qp`).

````sage
from sage.all import Qp

# Definimos el campo de los números 5-ádicos con una precisión de 10 dígitos
Q5 = Qp(5, prec=10)

print("--- Representación de enteros negativos ---")
print(f"-1 en Q_5: {Q5(-1)}")

print("\n--- Representación de números racionales ---")
# 1/(1 - p) = 1/(1 - 5) = -1/4
print(f"-1/4 en Q_5: {Q5(-1/4)}")

# Verifiquemos para otro primo, p = 7:
Q7 = Qp(7, prec=8)
print("\n--- Representación en Q_7 ---")
print(f"-1 en Q_7: {Q7(-1)}")
print(f"1/(1 - 7) = -1/6 en Q_7: {Q7(-1/6)}")
````

## La Valuación $p$-ádica

Profundicemos en la definición de la valuación $p$-ádica para enteros. La valuación es una función:
$$v_p: \mathbb{Z} \to \mathbb{Z}_{\geq 0} \cup \{\infty\}$$
que nos indica el exponente máximo de la potencia de $p$ que divide a un número $n$. Formalmente:
$$v_p(n) = \max \{ k \in \mathbb{Z}_{\geq 0} : p^k \mid n \}$$

Es decir, si expresamos un entero $n \neq 0$ de la forma:
$$n = p^k \cdot m \quad \text{con } \gcd(m, p) = 1$$
entonces la valuación de $n$ es exactamente $k$:
$$v_p(n) = k$$

### El caso de $n = 0$:
Por convención, definimos:
$$v_p(0) = \infty$$
Esto resulta natural si notamos que $0$ es divisible por cualquier potencia de $p$ (por ejemplo, $0 = 0 \cdot p^{2026}$). Así, el exponente de $p$ que divide a $0$ puede ser arbitrariamente grande.

### Ejemplos de valuación:
* $v_3(27) = 3$ (ya que $27 = 3^3$)
* $v_2(20) = v_2(2^2 \cdot 5) = 2$

Si extendemos la valuación a enteros negativos o fracciones que no tienen el factor $p$ en su numerador ni en su denominador, se tiene:
* $v_5(-1) = 0$ (pues $-1 = 4 + 4 \cdot 5 + 4 \cdot 5^2 + \dots$, que no comienza con un factor de 5).
* $v_p\left(\frac{1}{1-p}\right) = 0$ (ya que $\frac{1}{1-p} = 1 + p + p^2 + \dots$, cuyo primer término es 1, no divisible por $p$).

---

### Propiedades de la valuación $p$-ádica
Para cualesquiera $m, n \in \mathbb{Z}$ (y de hecho en $\mathbb{Q}$):
1. **Comportamiento logarítmico**:
   $$v_p(mn) = v_p(m) + v_p(n)$$
2. **Desigualdad ultramétrica**:
   $$v_p(m+n) \geq \min(v_p(m), v_p(n))$$
   *Intuitivamente, al sumar dos números, podemos factorizar la menor potencia de $p$ común a ambos, de modo que el resultado es divisible por al menos esa potencia.*

> **Nota**: Observe que estas propiedades hacen que $v_p$ se asemeje a un logaritmo con signo invertido. De hecho, el logaritmo usual se puede considerar heurísticamente como una "valuación en el infinito" (el caso arquimediano).

---

### Extensión de la valuación a $\mathbb{Q}$

Para extender la valuación a los racionales, escribimos cualquier racional $r \in \mathbb{Q}^*$ como $r = \frac{a}{b}$ con $a, b \in \mathbb{Z}$ y definimos:
$$v_p(r) = v_p\left(\frac{a}{b}\right) = v_p(a) - v_p(b)$$

Esta definición es independiente de la representación elegida de la fracción y preserva las propiedades 1 y 2.

````sage
from sage.all import ZZ, QQ

# Ejemplos de valuaciones
print(f"v_3(27) = {ZZ(27).valuation(3)}")
print(f"v_2(20) = {ZZ(20).valuation(2)}")
print(f"v_5(-1) = {ZZ(-1).valuation(5)}")
print(f"v_5(-1/4) = {QQ(-1/4).valuation(5)}")

# Extensión a racionales
r = QQ(3/20) # 3 / (2^2 * 5)
print(f"\nv_2(3/20) = {r.valuation(2)}")
print(f"v_5(3/20) = {r.valuation(5)}")
print(f"v_3(3/20) = {r.valuation(3)}")

# Verificación de la propiedad ultramétrica v_2(m+n) >= min(v_2(m), v_2(n))
m = QQ(3/4)
n = QQ(5/2)
val_m = m.valuation(2)
val_n = n.valuation(2)
val_sum = (m + n).valuation(2)

print(f"\nVerificación propiedad ultramétrica para m = {m}, n = {n} (p = 2):")
print(f"v_2({m}) = {val_m}")
print(f"v_2({n}) = {val_n}")
print(f"v_2({m} + {n}) = v_2({m+n}) = {val_sum}")
print(f"¿Se cumple v_2(m+n) >= min(v_2(m), v_2(n))?: {val_sum >= min(val_m, val_n)}")
````

# Métrica inducida por la valuación $p$-ádica

## El Valor Absoluto $p$-ádico

Cada primo $p$ induce un valor absoluto (o norma) en los números racionales $\mathbb{Q}$, definido como:
$$|r|_p = p^{-v_p(r)} \quad \text{para } r \neq 0$$

### El caso de $r = 0$:
Dado que $v_p(0) = \infty$, intuitivamente se tiene:
$$|0|_p = p^{-\infty} = 0$$
Esto puede formalizarse como el límite de $p^{-k}$ cuando $k \to \infty$, lo cual es consistente con la definición usual de que la norma de cero debe ser cero.

### Sobre la base del valor absoluto:
La elección de la base $1/p$ (es decir, $p^{-v_p(r)}$) es convencional y estética. Podríamos haber definido el valor absoluto usando cualquier otra base $\alpha > 1$ como $\alpha^{-v_p(r)}$. Sin embargo, se elige precisamente $p$ por razones profundas que se relacionan con la fórmula del producto global (Teorema de Ostrowski), la cual conecta todos los valores absolutos posibles sobre $\mathbb{Q}$.

### Interpretación semántica de la distancia al cero:
El valor absoluto $p$-ádico mide **qué tan divisible es un número por $p$**:
* Los números **altamente divisibles por $p$** tienen una valuación grande y positiva, lo que hace que su valor absoluto $p$-ádico sea **muy pequeño (cercano a cero)**.
* Los números **cuyo denominador es divisible por $p$** tienen una valuación negativa, lo que hace que su valor absoluto $p$-ádico sea **muy grande (lejos de cero)**.

#### Ejemplo con $p = 3$:
* $|27|_3 = 3^{-3} = \frac{1}{27}$
* $|9|_3 = 3^{-2} = \frac{1}{9}$

Dado que $\frac{1}{27} < \frac{1}{9}$, el número $27$ es **más pequeño** y está **más cerca del cero** que el número $9$ en la métrica 3-ádica.

---

### Más ejemplos semánticos para otros primos:

#### 1. Caso $p = 5$ (Divisibilidad y cercanía al cero)
Consideremos los números $3$, $50$ y $250$:
* **Para $3$**: $3$ no es divisible por 5. Su valuación es $v_5(3) = 0 \implies |3|_5 = 5^0 = 1$.
* **Para $50$**: $50 = 2 \cdot 5^2$. Su valuación es $v_5(50) = 2 \implies |50|_5 = 5^{-2} = \frac{1}{25}$.
* **Para $250$**: $250 = 2 \cdot 5^3$. Su valuación es $v_5(250) = 3 \implies |250|_5 = 5^{-3} = \frac{1}{125}$.

**Explicación semántica**: 
En la métrica 5-ádica, el número $250$ está más cerca del cero que $50$ porque tiene mayor grado de divisibilidad por 5 ($|250|_5 = 1/125 < |50|_5 = 1/25$). Por otro lado, $3$ no es divisible por 5 en absoluto, por lo que su tamaño 5-ádico es el máximo posible para un entero, situándose bastante lejos del cero en comparación.

#### 2. Caso $p = 2$ (Fracciones y tamaño grande)
Consideremos los números $64$, $12$ y $\frac{1}{8}$:
* **Para $64$**: $64 = 2^6$. Su valuación es $v_2(64) = 6 \implies |64|_2 = 2^{-6} = \frac{1}{64}$.
* **Para $12$**: $12 = 3 \cdot 2^2$. Su valuación es $v_2(12) = 2 \implies |12|_2 = 2^{-2} = \frac{1}{4}$.
* **Para $\frac{1}{8}$**: $\frac{1}{8} = 2^{-3}$. Su valuación es $v_2\left(\frac{1}{8}\right) = -3 \implies |\frac{1}{8}|_2 = 2^{-(-3)} = 2^3 = 8$.

**Explicación semántica**:
Aquí vemos que $64$ es extremadamente pequeño ($1/64$) y está muy cerca de cero, mientras que $12$ está a una distancia intermedia ($1/4$). 
Sin embargo, la fracción $\frac{1}{8}$ es **enormemente grande** en el sentido 2-ádico ($|1/8|_2 = 8$), lo que significa que está muy lejos de cero. Esto se debe a que el factor 2 está en el denominador, representando una "divisibilidad negativa" que expande la distancia.

````sage
from sage.all import QQ

def abs_p_adico(x, p):
    """
    Calcula el valor absoluto p-ádico de x como un número racional de Sage.
    """
    x = QQ(x)
    if x == 0:
        return QQ(0)
    val = x.valuation(p)
    return QQ(p)**(-val)

# Pruebas con p = 5
print("--- Valores absolutos 5-ádicos (p=5) ---")
print(f"|3|_5   = {abs_p_adico(3, 5)}")
print(f"|50|_5  = {abs_p_adico(50, 5)}")
print(f"|250|_5 = {abs_p_adico(250, 5)}")

# Pruebas con p = 2
print("\n--- Valores absolutos 2-ádicos (p=2) ---")
print(f"|64|_2  = {abs_p_adico(64, 2)}")
print(f"|12|_2  = {abs_p_adico(12, 2)}")
print(f"|1/8|_2 = {abs_p_adico(1/8, 2)}")
````

Con la definición del valor absoluto $p$-ádico, es inmediato ver que:
$$\lim_{n \to \infty} |p^n|_p = \lim_{n \to \infty} p^{-n} = 0$$
Es decir, las potencias grandes de primos se hacen "chiquititas" en esta métrica.

Le damos el nombre de **valor absoluto** precisamente porque satisface todas las propiedades axiomáticas de un valor absoluto clásico, e incluso posee propiedades más fuertes.

---

### Teorema: Propiedades del Valor Absoluto $p$-ádico

Para todo $x, y \in \mathbb{Q}$, se cumple que:
1. $|x|_p = 0 \iff x = 0$
2. $|xy|_p = |x|_p |y|_p$ (Multiplicatividad)
3. $|x+y|_p \leq \max(|x|_p, |y|_p)$ (**Desigualdad ultramétrica**)

#### Demostración:
1. **Definida positiva**: $|x|_p = p^{-v_p(x)}$. Si $x \neq 0$, $v_p(x)$ es finito, luego $p^{-v_p(x)} > 0$. Si $x = 0$, $v_p(0) = \infty$, luego $p^{-\infty} = 0$. Por lo tanto, $|x|_p = 0 \iff x = 0$.
2. **Multiplicatividad**: Usando la propiedad logarítmica de la valuación $v_p(xy) = v_p(x) + v_p(y)$, tenemos:
   $$|xy|_p = p^{-v_p(xy)} = p^{-(v_p(x) + v_p(y))} = p^{-v_p(x)} p^{-v_p(y)} = |x|_p |y|_p$$
3. **Desigualdad ultramétrica**: Usando la propiedad $v_p(x+y) \geq \min(v_p(x), v_p(y))$, multiplicamos por $-1$ y obtenemos:
   $$-v_p(x+y) \leq -\min(v_p(x), v_p(y)) = \max(-v_p(x), -v_p(y))$$
   Como la función exponencial $f(t) = p^t$ es estrictamente creciente (ya que $p > 1$), al aplicarla a ambos lados se preserva la desigualdad:
   $$p^{-v_p(x+y)} \leq p^{\max(-v_p(x), -v_p(y))} = \max(p^{-v_p(x)}, p^{-v_p(y)})$$
   Lo cual es exactamente:
   $$|x+y|_p \leq \max(|x|_p, |y|_p)$$
   $\blacksquare$

---

### Conexión con la propiedad Arquimediana

La desigualdad ultramétrica es una condición mucho **más fuerte** que la desigualdad triangular clásica ($|x+y| \leq |x| + |y|$). Esta propiedad hace que el valor absoluto $p$-ádico sea **no arquimediano**.

**¿Qué significa ser (o no ser) arquimediano?**
En un valor absoluto arquimediano (como el valor absoluto usual en $\mathbb{R}$), si tomamos el número $1$ y lo sumamos consigo mismo múltiples veces, su tamaño crece indefinidamente:
$$|1 + 1 + \dots + 1|_{\infty} = |n|_{\infty} = n \to \infty$$

Sin embargo, en el valor absoluto $p$-ádico, al usar la desigualdad ultramétrica de forma reiterada obtenemos:
$$|n|_p = |1 + 1 + \dots + 1|_p \leq \max(|1|_p, |1|_p, \dots, |1|_p) = 1$$
¡No importa cuántas veces sumemos $1$ consigo mismo, su valor absoluto $p$-ádico **nunca** excederá 1! Esto viola fundamentalmente el principio de Arquímedes, y por ello decimos que $|\cdot|_p$ es una norma **no arquimediana**.

---

### Métrica Inducida

Dado que $|\cdot|_p$ satisface todas las propiedades de un valor absoluto, induce de manera natural una distancia (métrica) en los números racionales $\mathbb{Q}$:
$$d_p(x, y) = |x - y|_p$$

Debido a la desigualdad ultramétrica, esta distancia satisface una versión fuerte de la desigualdad triangular:
$$d_p(x, z) \leq \max(d_p(x, y), d_p(y, z))$$
Esto da lugar a una topología fascinante, la geometría ultramétrica, donde, por ejemplo, todos los triángulos son isósceles y cualquier punto dentro de una bola puede considerarse el centro de la misma.

# Completando el espacio: Los Enteros $p$-ádicos ($\mathbb{Z}_p$)

Si consideramos el conjunto de los números naturales $\mathbb{N}$ (o incluso $\mathbb{Z}$) dotado de la métrica $d_p(x, y) = |x - y|_p$, resulta que **no es un espacio métrico completo**.

### Ejemplo de una sucesión de Cauchy que no converge en $\mathbb{N}$
Consideremos la sucesión de sumas parciales $x_n = \sum_{k=0}^{n-1} p^k = 1 + p + p^2 + \dots + p^{n-1}$. 
Es evidente que $x_n \in \mathbb{N}$ para todo $n \geq 1$. 
Para ver que es una sucesión de Cauchy en la métrica $p$-ádica, tomemos $m > n$:
$$|x_m - x_n|_p = |\sum_{k=n}^{m-1} p^k|_p = |p^n(1 + p + \dots + p^{m-n-1})|_p$$
Usando la multiplicatividad y sabiendo que el término entre paréntesis es un entero (cuya norma es $\leq 1$):
$$|x_m - x_n|_p \leq |p^n|_p = p^{-n}$$
Cuando $n, m \to \infty$, la distancia $p^{-n} \to 0$. Por lo tanto, $(x_n)$ es una sucesión de Cauchy.
Sin embargo, como vimos en la sección anterior, el límite formal de esta serie es $\frac{1}{1-p}$. Si $p=5$, el límite es $-1/4$. ¡Este límite no es un número natural (ni siquiera es un entero)! Por tanto, esta sucesión de Cauchy converge a algo que está fuera de $\mathbb{N}$ y de $\mathbb{Z}$.

---

### La Completación $\mathbb{Z}_p$

Al igual que completamos $\mathbb{Q}$ con la métrica usual para obtener los números reales $\mathbb{R}$ (donde todo decimal infinito tiene sentido), si completamos $\mathbb{N}$ (o $\mathbb{Q}$) respecto a la métrica $d_p$, obtenemos un nuevo espacio.

Al completar, garantizamos que todas estas sumas infinitas formales converjan. El conjunto resultante es el de las series infinitas de potencias de $p$:
$$\overline{\mathbb{N}} = \left\lbrace \sum_{k=0}^{\infty} a_k p^k : a_k \in \{0, 1, \dots, p-1\} \right\rbrace$$

Este espacio de completación se conoce como el anillo de los **enteros $p$-ádicos**, y se denota como **$\mathbb{Z}_p$**.

### ¿Contiene $\mathbb{Z}_p$ a todos los enteros?
En una primera impresión, $\mathbb{Z}_p$ podría parecer que solo contiene números "positivos", ya que está formado por combinaciones de coeficientes $a_k \geq 0$. Podríamos dudar de que contenga a los números negativos y, por consiguiente, a todo $\mathbb{Z}$.

Pero la respuesta es que **sí los contiene**. La clave está en nuestra explicación previa de que $-1$ se puede expresar como la serie infinita de $(p-1)$'s:
$$-1 = (\dots(p-1)(p-1)(p-1))_p = \sum_{k=0}^{\infty} (p-1) p^k$$
Como esta serie infinita es un elemento válido de $\mathbb{Z}_p$ (pues sus coeficientes están en $\{0, \dots, p-1\}$), entonces $-1 \in \mathbb{Z}_p$. Y al tener a los números naturales y al $-1$, tenemos también a todos los enteros. Por lo tanto, $\mathbb{Z} \subset \mathbb{Z}_p$ de forma natural.

### Proposición: $\mathbb{Z}$ es denso en $\mathbb{Z}_p$

Una consecuencia directa de nuestra construcción es que los enteros $\mathbb{Z}$ (y por extensión $\mathbb{N}$) forman un subconjunto **denso** dentro de $\mathbb{Z}_p$. Es decir, la clausura de $\mathbb{Z}$ bajo la métrica $p$-ádica es todo $\mathbb{Z}_p$:
$$\overline{\mathbb{Z}} = \mathbb{Z}_p$$

Cualquier entero $p$-ádico puede ser aproximado con la precisión deseada por un entero clásico simplemente truncando su serie infinita $p$-ádica.

---

### Topología y Geometría de $\mathbb{Z}_p$

De hecho, $\mathbb{Z}_p$ coincide exactamente con la **bola cerrada unitaria** (la bola de radio 1 con centro en el origen) dentro del espacio métrico:
$$\mathbb{Z}_p = \{ x \in \mathbb{Q}_p : |x|_p \leq 1 \}$$

Las "bolas" en los números $p$-ádicos tienen propiedades extremadamente anti-intuitivas debido a la naturaleza **no arquimediana** (desigualdad ultramétrica) del espacio. Algunas de estas propiedades exóticas incluyen:

- **Todo punto dentro de una bola es su centro**: Si tomas cualquier punto dentro de una bola $p$-ádica y trazas una bola del mismo radio alrededor de él, obtienes exactamente la misma bola original.
- **Bolas disjuntas o contenidas**: Dadas dos bolas, o bien son completamente disjuntas, o una está completamente contenida dentro de la otra. ¡No pueden intersectarse parcialmente!
- **Todos los triángulos son isósceles**: Si tomas tres puntos cualesquiera, al menos dos de las distancias entre ellos son exactamente iguales (y la tercera es menor o igual).

Es una geometría muy curiosa.

---

### ¿Cuáles otros números están por ahí metidos?

Ya vimos que los naturales, los negativos (como el -1) y ciertas fracciones que no tienen denominadores divisibles por $p$ (como $\frac{1}{1-p}$) forman parte de $\mathbb{Z}_p$. Pero esto nos lleva de manera natural a preguntarnos:
**¿Qué otros tipos de números están escondidos dentro de estas expansiones $p$-ádicas infinitas?**

---
### Proposición: Números racionales en $\mathbb{Z}_p$

Para cualquier número racional $r \in \mathbb{Q}$, si podemos escribirlo de la forma $r = p^k \frac{n}{d}$ donde $\gcd(n, d) = 1$, $\gcd(n, p) = 1$, $\gcd(d, p) = 1$ (es decir, todas las partes coprimas) y $k \geq 0$, entonces vale que $r \in \mathbb{Z}_p$.

**Ejemplo**: $1/2 \in \mathbb{Z}_3$. Escribimos $3^0 \cdot 1/2 = 1/2$. 
Por el teorema de Bachet-Bézout, existen enteros $u, t$ tales que $2u + 3t = 1$. Tomando $(u, t) = (2, -1)$:
$$\frac{1}{2} = \frac{1}{2}(2(2) + 3(-1)) = 2 + 3\left(\frac{-1}{2}\right)$$
Nuevamente por Bachet-Bézout para el numerador $-1$, buscamos $2u + 3t = -1$. Con $(u, t) = (1, -1)$ obtenemos $2(1) + 3(-1) = -1$, luego:
$$\frac{-1}{2} = \frac{1}{2}(2(1) + 3(-1)) = 1 + 3\left(\frac{-1}{2}\right)$$
Al sustituir recursivamente, nos damos cuenta de que el término $\frac{-1}{2}$ vuelve a aparecer, generando un ciclo infinito de $1$s:
$$\frac{1}{2} = 2 + 3(1 + 3(1 + 3(\dots))) = (\dots 11112)_3$$

---

**Otro Ejemplo**: $\frac{1}{3} \in \mathbb{Z}_5$. 
Como $\gcd(3, 5) = 1$, la fracción tiene valuación positiva (es cero) y pertenece a los enteros 5-ádicos. Apliquemos Bachet-Bézout buscando $3u + 5t = 1$. Notemos que $3(2) + 5(-1) = 1$.
$$\frac{1}{3} = \frac{1}{3}(3(2) + 5(-1)) = 2 + 5\left(\frac{-1}{3}\right)$$
El primer dígito es $a_0 = 2$. Ahora desarrollamos $\frac{-1}{3}$. Buscamos $3u + 5t = -1$. Notamos que $3(3) + 5(-2) = -1$.
$$\frac{-1}{3} = \frac{1}{3}(3(3) + 5(-2)) = 3 + 5\left(\frac{-2}{3}\right)$$
El dígito es $a_1 = 3$. Ahora desarrollamos $\frac{-2}{3}$. Buscamos $3u + 5t = -2$. Notamos que $3(1) + 5(-1) = -2$.
$$\frac{-2}{3} = \frac{1}{3}(3(1) + 5(-1)) = 1 + 5\left(\frac{-1}{3}\right)$$
El dígito es $a_2 = 1$. ¡Pero hemos vuelto a $\frac{-1}{3}$! Esto significa que los dígitos entrarán en un ciclo $3, 1, 3, 1, \dots$ hacia el infinito. Por tanto:
$$\frac{1}{3} = (\dots 1313132)_5$$

**Verificación**: 
Multipliquemos este número $5$-ádico por $3$ para verificar que obtenemos $1$:
* $3 \times 2 = 6 = 1 + 5(1)$ $\implies$ dígito **1**, llevamos 1.
* $3 \times 3 + 1 = 10 = 0 + 5(2)$ $\implies$ dígito **0**, llevamos 2.
* $3 \times 1 + 2 = 5 = 0 + 5(1)$ $\implies$ dígito **0**, llevamos 1.
* $3 \times 3 + 1 = 10 = 0 + 5(2)$ $\implies$ dígito **0**, llevamos 2.
Como el patrón se repite, arrastramos ceros indefinidamente. El resultado es efectivamente $(\dots 00001)_5 = 1$.

### Otra metodología: Uso de Series Geométricas y el Teorema de Euler

Existe otra forma muy elegante de encontrar la expansión $p$-ádica de una fracción $1/d$, aprovechando que $p^k \equiv 1 \pmod d$ para algún $k$ (garantizado por el Teorema de Euler si $\gcd(p, d) = 1$).

**Ejemplo de $1/7 \in \mathbb{Z}_3$:**
1. Sabemos que $3^6 \equiv 1 \pmod 7$, ya que $3^6 - 1 = 729 - 1 = 728 = 7 \times 104$.
2. Podemos despejar $1/7$:
   $$\frac{1}{7} = \frac{104}{3^6 - 1} = \frac{-104}{1 - 3^6}$$
3. Sabemos que $\frac{1}{1 - 3^6} = 1 + 3^6 + 3^{12} + \dots$ es una serie geométrica. Por lo tanto:
   $$\frac{1}{7} = -104 (1 + 3^6 + 3^{12} + \dots)$$
   Pero $-104$ no es un "dígito" válido en base $3^6$. Para arreglarlo, escribimos $-104$ como un número positivo menos $3^6$:
   $$-104 = 729 - 104 - 3^6 = 625 - 3^6 = 624 + (1 - 3^6)$$
4. Sustituimos esto en la fracción:
   $$\frac{1}{7} = \frac{624 + (1 - 3^6)}{1 - 3^6} = 1 + \frac{624}{1 - 3^6}$$
5. Ahora $624$ es positivo y $0 \leq 624 < 3^6$. Lo convertimos a base 3 (necesitamos 6 dígitos):
   $$624 = 2 \cdot 3^5 + 1 \cdot 3^4 + 2 \cdot 3^3 + 0 \cdot 3^2 + 1 \cdot 3^1 + 0 \cdot 3^0 = (212010)_3$$
6. La fracción $\frac{624}{1 - 3^6}$ representa repetir el bloque $212010$ infinitamente:
   $$\frac{624}{1 - 3^6} = (\overline{212010})_3$$
7. Finalmente, sumamos el $1$ libre que nos quedó:
   $$\frac{1}{7} = 1 + (\overline{212010})_3 = (\dots 212010212011)_3$$

---

**Nuevo Ejemplo con esta metodología: $1/5 \in \mathbb{Z}_2$**
1. **Buscar $k$**: Necesitamos $2^k \equiv 1 \pmod 5$. Vemos que $2^4 = 16 \equiv 1 \pmod 5$. Por lo tanto, $k=4$.
2. **Escribir la fracción**: Como $2^4 - 1 = 15 = 5 \times 3$, tenemos:
   $$\frac{1}{5} = \frac{3}{2^4 - 1} = \frac{-3}{1 - 2^4}$$
3. **Ajustar el numerador**: Expresamos $-3$ en términos de $1 - 2^4$ (donde $2^4 = 16$):
   $$-3 = 16 - 3 - 16 = 13 - 16 = 12 + (1 - 2^4)$$
   Sustituyendo:
   $$\frac{1}{5} = \frac{12 + (1 - 2^4)}{1 - 2^4} = 1 + \frac{12}{1 - 2^4}$$
4. **Convertir a base 2**: Expresamos $12$ en base 2 con $k=4$ dígitos:
   $$12 = 8 + 4 = 1 \cdot 2^3 + 1 \cdot 2^2 + 0 \cdot 2^1 + 0 \cdot 2^0 = (1100)_2$$
5. **Construir el número**: La serie $\frac{12}{1 - 2^4}$ es el bloque repetido infinitamente:
   $$\frac{12}{1 - 2^4} = (\overline{1100})_2$$
   Por último, sumamos $1$:
   $$\frac{1}{5} = 1 + (\overline{1100})_2 = (\dots 110011001101)_2$$

*(¡Puedes comprobar que al multiplicar esto por $5 = (101)_2$ con la suma binaria infinita, el resultado es exactamente $1$!)*

---

### El Cuerpo de Fracciones $\mathbb{Q}_p$

**Ejemplo de lo que NO está**: $p^{-k}$ (como $1/p$ o $1/p^2$) no está en $\mathbb{Z}_p$, ya que su valuación $p$-ádica es estrictamente negativa ($-k$), lo que significa que su norma es estrictamente mayor a 1, saliéndose de la "bola" $\mathbb{Z}_p$.

Esto nos motiva a agregar términos de la forma $1/p^k$. Ya que $\mathbb{Z}_p$ es un dominio entero, podemos construir su cuerpo de fracciones:
$$\mathbb{Q}_p = \text{Frac}(\mathbb{Z}_p)$$
A este conjunto lo llamamos el **cuerpo de los números $p$-ádicos**. Los elementos de $\mathbb{Q}_p$ no son solo series de potencias positivas, sino que admiten un número **finito** de términos con potencias negativas, asemejándose a las **series de Laurent**:
$$\sum_{k=-N}^{\infty} a_k p^k \quad \text{con } a_k \in \{0, \dots, p-1\}$$
Esto completa la construcción de $\mathbb{Q}_p$ tanto métrica como algebraicamente.

# Sesión 3: Discriminantes

En esta sesión formalizaremos la definición de **discriminante** y exploraremos su profunda conexión con la ramificación de ideales primos, el teorema de Minkowski, la teoría de Kronecker-Weber y la fórmula del conductor-discriminante.

## Definición del Discriminante

Sea $L/K$ una extensión de Galois de grado $n$ y $\operatorname{Gal}(L/K) = \{\sigma_1, \dots, \sigma_n\}$. Si $\alpha_1, \dots, \alpha_n \in \mathcal{O}_L$, definimos el **discriminante** de esta $n$-tupla como:

$$d(\alpha_1, \dots, \alpha_n) = \det\Big( \sigma_i(\alpha_j) \Big)^2$$

El **ideal discriminante** de la extensión $L/K$, denotado como $\operatorname{disc}(L/K)$, es el ideal de $\mathcal{O}_K$ generado por los discriminantes de todas las posibles $n$-tuplas de elementos en el anillo de enteros $\mathcal{O}_L$:

$$\operatorname{disc}(L/K) = \langle d(\mathbf{lpha}) : \mathbf{lpha} \in \mathcal{O}_L^n \rangle_K$$

### El caso global $L/\mathbb{Q}$

Cuando el cuerpo base es $K = \mathbb{Q}$, el anillo de enteros $\mathcal{O}_L$ es un $\mathbb{Z}$-módulo libre de rango $n$. Por lo tanto, posee una **base de enteros** (o base integral) $\mathcal{O}_L = \mathbb{Z}\omega_1 + \dots + \mathbb{Z}\omega_n$. 

En este caso, el ideal $\operatorname{disc}(L/\mathbb{Q})$ es principal y está generado por el discriminante de dicha base de enteros. Por comodidad, definimos el discriminante de $L$ directamente como este valor entero:

$$\operatorname{disc}(L) = \operatorname{disc}(L/\mathbb{Q}) = d(\omega_1, \dots, \omega_n)$$

**Ejemplo ($L = \mathbb{Q}(i)$):**
El anillo de enteros es $\mathcal{O}_L = \mathbb{Z}\cdot 1 + \mathbb{Z}\cdot i$. Los dos automorfismos de $\operatorname{Gal}(L/\mathbb{Q})$ son la identidad y la conjugación compleja. Calculamos:

$$\operatorname{disc}(L) = \det \begin{pmatrix} 1 & i \\ 1 & -i \end{pmatrix}^2 = (-2i)^2 = -4$$

### Signo del Discriminante y Teorema de Brill

El signo del discriminante de un cuerpo de números $L$ está determinado por su **firma** $(r_1, r_2)$, donde $r_1$ es el número de inmersiones reales y $r_2$ es el número de parejas de inmersiones complejas conjugadas ($n = r_1 + 2r_2$). El **Teorema de Brill** establece que:

$$\operatorname{signo}(\operatorname{disc}(L)) = (-1)^{r_2}$$

**Ejemplos de signos:**
1. **Cuerpos cuadráticos reales**: Para $L = \mathbb{Q}(\sqrt{d})$ con $d > 0$ libre de cuadrados, las dos inmersiones son reales ($r_1=2, r_2=0$), por lo que $\operatorname{disc}(L) > 0$. 
   * Para $\mathbb{Q}(\sqrt{5})$ con $5 \equiv 1 \pmod 4$, la base de enteros es $\{1, \frac{1+\sqrt{5}}{2}\}$. Obtenemos:
     $$\operatorname{disc}(L) = \det \begin{pmatrix} 1 & \frac{1+\sqrt{5}}{2} \\ 1 & \frac{1-\sqrt{5}}{2} \end{pmatrix}^2 = (-\sqrt{5})^2 = 5 > 0$$
   * Para $\mathbb{Q}(\sqrt{2})$ con $2 \equiv 2 \pmod 4$, la base de enteros es $\{1, \sqrt{2}\}$. Obtenemos:
     $$\operatorname{disc}(L) = \det \begin{pmatrix} 1 & \sqrt{2} \\ 1 & -\sqrt{2} \end{pmatrix}^2 = (-2\sqrt{2})^2 = 8 > 0$$
2. **Cuerpos cuadráticos imaginarios**: Para $L = \mathbb{Q}(\sqrt{d})$ con $d < 0$ libre de cuadrados, las dos inmersiones son complejas conjugadas ($r_1=0, r_2=1$), por lo que $\operatorname{disc}(L) < 0$. El ejemplo de $\mathbb{Q}(i)$ tiene $\operatorname{disc}(\mathbb{Q}(i)) = -4 < 0$.

## Teorema de Minkowski y Ramificación Global

Un resultado de gran peso en la geometría de números de Minkowski establece que para cualquier cuerpo de números algebraicos $L \neq \mathbb{Q}$ de grado $n$:

$$|\operatorname{disc}(L)| > 1$$

Una consecuencia inmediata de este teorema, combinada con la teoría de ramificación, es que **toda extensión propia $L/\mathbb{Q}$ está ramificada** en al menos un número primo. Este hecho dista de ser obvio y marca una diferencia fundamental con las extensiones de curvas algebraicas o superficies de Riemann, donde sí existen revestimientos no ramificados de grado mayor a uno.

---

## El Discriminante como Detector de Ramificación

El discriminante no solo mide el volumen reticular del anillo de enteros, sino que codifica de forma exacta qué primos se ramifican en la extensión.

**Teorema:** Sea $L/K$ una extensión de Galois de cuerpos de números y sea $\mathfrak{p} \subseteq \mathcal{O}_K$ un ideal primo. Entonces:

$$\mathfrak{p} \text{ ramifica en } L \iff \mathfrak{p} \mid \operatorname{disc}(L/K)$$

donde la relación de división $\mid$ se entiende como división de ideales en $\mathcal{O}_K$.

### ¿Por qué se ramifica una cantidad finita de primos?

Este teorema nos permite obtener una conclusión crucial:
> **Corolario:** En cualquier extensión finita $L/K$ de cuerpos de números algebraicos, existe únicamente una **cantidad finita** de ideales primos $\mathfrak{p} \subseteq \mathcal{O}_K$ que se ramifican.

**Explicación/Demostración:**
1. Dado que la extensión $L/K$ es separable, el ideal discriminante $\operatorname{disc}(L/K)$ es un **ideal no nulo** de $\mathcal{O}_K$.
2. Como el anillo de enteros $\mathcal{O}_K$ es un **dominio de Dedekind**, todo ideal no nulo admite una factorización única en factores primos de la forma:
   $$\operatorname{disc}(L/K) = \mathfrak{p}_1^{a_1} \mathfrak{p}_2^{a_2} \dots \mathfrak{p}_m^{a_m}$$
   donde los exponentes $a_i$ son enteros positivos y la lista de ideales primos $\{\mathfrak{p}_1, \dots, \mathfrak{p}_m\}$ es finita.
3. Por definición, un ideal primo $\mathfrak{p}$ divide a $\operatorname{disc}(L/K)$ si y sólo si $\mathfrak{p}$ es uno de los factores primos de la factorización anterior (es decir, $\mathfrak{p} \in \{\mathfrak{p}_1, \dots, \mathfrak{p}_m\}$).
4. En consecuencia, los únicos ideales primos que se ramifican en $L$ son aquellos que pertenecen a este conjunto finito. $\quad\blacksquare$

## Teorema de Kronecker-Weber y el Conductor

El **Teorema de Kronecker-Weber** clasifica las extensiones abelianas de $\mathbb{Q}$:

**Teorema:** Si $L/\mathbb{Q}$ es una extensión abeliana finita (es decir, Galois con grupo de Galois abeliano), entonces existe un número entero $n \in \mathbb{N}$ tal que:

$$L \subseteq \mathbb{Q}(\zeta_n)$$

donde $\zeta_n = e^{2\pi i / n}$ es una raíz $n$-ésima primitiva de la unidad.

### El Conductor
El mínimo entero positivo $n$ para el cual se cumple la contención anterior se conoce como el **conductor** de $L$, denotado comúnmente como $f(L)$ (o simplemente $f$).

El conductor $f$ está íntimamente ligado a la ramificación: un número primo $p$ se ramifica en $L$ si y sólo si $p \mid f$. De hecho, el conductor tiene una factorización de la forma:

$$f = \prod_{p \text{ ramifica en } L/\mathbb{Q}} p^{n_p}$$

donde los exponentes $n_p$ están relacionados con la ramificación salvaje del primo $p$.

### ¿Qué significa la "n" en el escenario local $\mathbb{Q}_p$?

En el contexto local, el teorema de Kronecker-Weber también es válido:
> **Teorema (Kronecker-Weber local):** Si $K/\mathbb{Q}_p$ es una extensión abeliana finita, entonces existe un entero positivo $n \in \mathbb{N}$ tal que $K \subseteq \mathbb{Q}_p(\zeta_n)$, donde $\mathbb{Q}_p(\zeta_n)$ es el cuerpo obtenido al adjuntar una raíz $n$-ésima primitiva de la unidad $\zeta_n$ a $\mathbb{Q}_p$.

**Explicación de la $n$ en el caso local:**
* La $n$ local sigue siendo un **entero positivo clásico**. Sin embargo, la estructura del cuerpo ciclotómico local $\mathbb{Q}_p(\zeta_n)$ se puede descomponer según la divisibilidad de $n$ por $p$. Escribimos $n = m \cdot p^a$ con $\gcd(m, p) = 1$:
  1. La extensión $\mathbb{Q}_p(\zeta_m)$ es **no ramificada** (unramified) sobre $\mathbb{Q}_p$, y su grado es el orden multiplicativo de $p$ módulo $m$.
  2. La extensión $\mathbb{Q}_p(\zeta_{p^a})$ es **totalmente ramificada** (totally ramified) sobre $\mathbb{Q}_p$, de grado $\phi(p^a) = p^{a-1}(p-1)$.
* Así, la $n$ local codifica tanto la parte no ramificada (el factor $m$ coprimo con $p$) como la parte ramificada (la potencia $p^a$) del cuerpo abeliano local.

## Fórmula del Conductor-Discriminante

Por la teoría de Galois de extensiones abelianas y el teorema de Kronecker-Weber, el grupo de Galois $G = \operatorname{Gal}(L/\mathbb{Q})$ es isomorfo a un cociente de $(\mathbb{Z}/n\mathbb{Z})^\times$. Los **caracteres** de $G$ son homomorfismos:

$$\chi : G \to \mathbb{C}^\times$$

Cada uno de estos caracteres $\chi$ induce un carácter de Dirichlet (que denotamos también $\chi$) de módulo $n$. El **conductor del carácter**, denotado $f_\chi$, es el mínimo entero positivo para el cual $\chi$ se factoriza a través de $(\mathbb{Z}/f_\chi\mathbb{Z})^\times$.

Esto nos da un diagrama conmutativo de proyecciones:

$$
\begin{array}{ccc}
\operatorname{Gal}(L/\mathbb{Q}) & \longrightarrow & (\mathbb{Z}/f_\chi\mathbb{Z})^\times \\
& \searrow & \downarrow \\
& & (\mathbb{Z}/n\mathbb{Z})^\times
\end{array}
$$

La **Fórmula del Conductor-Discriminante de Hasse** establece una relación profunda entre el discriminante global y los conductores de sus caracteres individuales:

$$\operatorname{disc}(L) = \prod_{\chi} f_\chi$$

donde el producto corre sobre todos los caracteres $\chi$ del grupo de Galois $\operatorname{Gal}(L/\mathbb{Q})$ (entendido como ideales o directamente como valores enteros positivos).

   Tu intuición sobre el determinante de Vandermonde es excelente. Veamos por qué el signo $(-1)^{\frac{p-1}{2}}$ aparece y no se cancela con los cuadrados.

   La matriz de Vandermonde se construye con las raíces de $f(x) = \frac{x^p-1}{x-1} = x^{p-1} + x^{p-2} + \dots + 1$, que son $\zeta_p, \zeta_p^2, \dots, \zeta_p^{p-1}$.

   $$ \Delta = \prod_{1 \le i < j \le p-1} (\zeta_p^i - \zeta_p^j)^2 $$
   Dado que las raíces son números complejos (no reales), las diferencias $\zeta_p^i - \zeta_p^j$ son números complejos, por lo que sus cuadrados **no son necesariamente números reales positivos**. De hecho, hay una fórmula general muy útil que relaciona este producto con la norma de la derivada:
   $$ \prod_{1 \le i < j \le n} (\alpha_i - \alpha_j)^2 = (-1)^{\frac{n(n-1)}{2}} \prod_{i=1}^n f'(\alpha_i) = (-1)^{\frac{n(n-1)}{2}} N_{K/\mathbb{Q}}(f'(\alpha)) $$

   Partimos de $(x-1)f(x) = x^p - 1$. Derivando a ambos lados obtenemos:
   $$ f(x) + (x-1)f'(x) = p x^{p-1} $$

   $$ (\zeta_p - 1)f'(\zeta_p) = p \zeta_p^{p-1} $$

   $$ \operatorname{disc}(K) = (-1)^{\frac{p-1}{2}} N_{K/\mathbb{Q}}(f'(\zeta_p)) = (-1)^{\frac{p-1}{2}} p^{p-2} $$

   Por el teorema fundamental sobre ramificación, un número primo $q$ se ramifica en un cuerpo de números $K$ si y solo si $q$ divide a $\operatorname{disc}(K)$. 
   Como el discriminante es, salvo signo, una potencia de $p$ ($p^{p-2}$), el único primo que lo divide es $p$. Por lo tanto, **solo el primo $p$ se ramifica en $K$**.

# Sesión 4: Teoría de Cuerpos de Clase

Recordemos el teorema de Kummer-Dedekind:

Sea $L/K$ una extensión de cuerpos de números, finita y de Galois. Si $\mathfrak{p}$ es un ideal primo en $\mathcal{O}_K$, al extenderlo a $\mathcal{O}_L$ se descompone como:

$$\mathfrak{p}\mathcal{O}_L = \mathfrak{P}_1^e \mathfrak{P}_2^e \cdots \mathfrak{P}_g^e$$

Como la extensión es de Galois, todos los índices de ramificación son iguales ($e_i = e$), y todos los grados de inercia son iguales ($f_i = f$). Entonces tenemos la relación fundamental:

$$e \cdot f \cdot g = n$$

donde:
* $n = [L:K]$ es el grado de la extensión.
* $e$ es el **índice de ramificación**.
* $f = [\mathcal{O}_L/\mathfrak{P}_i : \mathcal{O}_K/\mathfrak{p}]$ es el **grado de inercia**.
* $g$ es el **grado de descomposición** (el número de factores primos distintos).

A veces trabajamos con anillos de enteros como $\mathbb{Z}[i]$ o $\mathbb{Z}[\sqrt{-5}]$ donde no se cumple la factorización única de elementos, pero **sí** se cumple la factorización única de ideales (ya que son dominios de Dedekind).

El objetivo de esta sesión es determinar cuáles primos $p \in \mathbb{Z}$ se pueden escribir de la forma $p = x^2 + 5y^2$. 

Existe una pregunta clásica muy parecida: ¿cuándo un primo se puede escribir como $p = x^2 + y^2$? Sabemos que esto es válido para $p = 2$ o si $p \equiv 1 \pmod 4$. Este problema no es tan difícil porque podemos factorizar la ecuación en $\mathbb{Z}[i]$ como $(x+iy)(x-iy) = p$. Como $\mathbb{Z}[i]$ es un Dominio de Ideales Principales (PID), la factorización única de elementos nos permite deducir el resultado fácilmente usando normas.

El problema con la ecuación $p = x^2 + 5y^2$ es que, al intentar factorizarla en el anillo $\mathbb{Z}[\sqrt{-5}]$, obtenemos:

$$(x + i\sqrt{5}y)(x - i\sqrt{5}y) = p$$

y $\mathbb{Z}[\sqrt{-5}]$ **no** es un PID (por ejemplo, $6 = 2 \cdot 3 = (1+\sqrt{-5})(1-\sqrt{-5})$).

Analicemos qué ocurre en el caso simple de $\mathbb{Z}[i]$:
* El discriminante del cuerpo $L = \mathbb{Q}(i)$ es $-4$. Por lo tanto, el único primo que ramifica es $2$, y de hecho ramifica como un cuadrado:
  $$(2)\mathcal{O}_L = \mathfrak{p}^2 = (1+i)^2\mathcal{O}_L \implies 1^2 + 1^2 = 2$$
* Por otro lado, si $p \equiv 1 \pmod 4$, el ideal se escinde (splits):
  $$(p)\mathcal{O}_L = \mathfrak{p}\mathfrak{q} = (x+iy)(x-iy)\mathcal{O}_L$$
  Tomando la norma, obtenemos $p = x^2 + y^2$. Todo esto funciona porque los ideales en $\mathbb{Z}[i]$ son principales.

Orientaremos nuestros esfuerzos a resolver $p = x^2 + 5y^2$ usando Teoría de Cuerpos de Clase.

Si $\mathfrak{p}$ es un ideal primo en $\mathcal{O}_K$, usando la descomposición de Kummer-Dedekind tenemos $\mathfrak{p}\mathcal{O}_L = (\mathfrak{P}_1 \cdots \mathfrak{P}_g)^e$. El grupo de Galois $G = \operatorname{Gal}(L/K)$ actúa transitivamente en el conjunto de primos $\{\mathfrak{P}_1, \dots, \mathfrak{P}_g\}$ sobre $\mathfrak{p}$. Es decir, para cualquier par de primos $\mathfrak{P}_i, \mathfrak{P}_j$, existe un $\sigma \in G$ tal que $\sigma(\mathfrak{P}_i) = \mathfrak{P}_j$.

Si $\mathfrak{P}$ es un primo sobre $\mathfrak{p}$, definimos su **grupo de descomposición** como:

$$D_{\mathfrak{P}} = \{\sigma \in G : \sigma(\mathfrak{P}) = \mathfrak{P}\}$$

Este grupo tiene orden $e \cdot f$, porque por el teorema de la órbita-estabilizador sabemos que el número de primos en la órbita es $g = |G|/|D_{\mathfrak{P}}| = n/|D_{\mathfrak{P}}| \implies |D_{\mathfrak{P}}| = e \cdot f$.

Esto es importante porque si $\sigma \in D_{\mathfrak{P}}$, dado que $\sigma \in \operatorname{Gal}(L/K)$, este automorfismo induce un automorfismo bien definido en el cuerpo residual $\mathcal{O}_L/\mathfrak{P}$ que fija $\mathcal{O}_K/\mathfrak{p}$. (No es inmediato que $\sigma$ preserve $\mathcal{O}_L$, pero basta notar que si $\alpha \in \mathcal{O}_L$ es raíz de un polinomio mónico con coeficientes en $\mathcal{O}_K$, entonces $\sigma(\alpha)$ también lo es, por lo que $\sigma(\mathcal{O}_L) \subseteq \mathcal{O}_L$).

Más aún, como $\sigma \in D_{\mathfrak{P}}$, podemos reducir este automorfismo al cuerpo residual $\mathcal{O}_L/\mathfrak{P}$. Llamemos a esta reducción $\bar{\sigma}$. Este mapeo entonces vive en $\operatorname{Gal}((\mathcal{O}_L/\mathfrak{P}) / (\mathcal{O}_K/\mathfrak{p}))$.

Tenemos un homomorfismo natural:
$$\phi: D_{\mathfrak{P}} \to \operatorname{Gal}((\mathcal{O}_L/\mathfrak{P}) / (\mathcal{O}_K/\mathfrak{p}))$$

El grupo de Galois residual tiene tamaño $f$ (el grado de inercia), y su kernel es el **grupo de inercia**:
$$I_{\mathfrak{P}} = \{\sigma \in D_{\mathfrak{P}} : \bar{\sigma} = \text{id}\}$$

el cual debe tener tamaño $e$ (el índice de ramificación). Así, obtenemos la sucesión exacta corta:
$$1 \to I_{\mathfrak{P}} \to D_{\mathfrak{P}} \to \operatorname{Gal}((\mathcal{O}_L/\mathfrak{P}) / (\mathcal{O}_K/\mathfrak{p})) \to 1$$

Como esta es una extensión de cuerpos finitos, el grupo de Galois $\operatorname{Gal}((\mathcal{O}_L/\mathfrak{P}) / (\mathcal{O}_K/\mathfrak{p}))$ es cíclico y está generado por un único elemento: el **automorfismo de Frobenius**. Dado que este elemento genera el grupo, tiene orden $f$.


En particular, por la sobreyectividad del homomorfismo $\phi$, existe un elemento $\sigma_{\mathfrak{P}} \in D_{\mathfrak{P}}$ (también denotado como $F_{\mathfrak{P}}$ o $\text{Frob}_{\mathfrak{P}}$) que se reduce al automorfismo de Frobenius residual en $\operatorname{Gal}((\mathcal{O}_L/\mathfrak{P}) / (\mathcal{O}_K/\mathfrak{p}))$. Si la extensión es no ramificada en $\mathfrak{P}$ (es decir, $e = 1$), este elemento es único y está unívocamente determinado por la condición:

$$\sigma_{\mathfrak{P}}(x) \equiv x^{N(\mathfrak{p})} \pmod{\mathfrak{P}} \quad \forall x \in \mathcal{O}_L$$

donde $N(\mathfrak{p}) = |\mathcal{O}_K/\mathfrak{p}|$ es el cardinal del cuerpo residual de la base.

Aquí es donde juega un rol fundamental la **abelianidad** de la extensión. Si consideramos otro primo $\mathfrak{P}'$ en $\mathcal{O}_L$ que yace sobre $\mathfrak{p}$, su elemento de Frobenius asociado $\sigma_{\mathfrak{P}'}$ es un conjugado en el grupo de Galois:

$$\tau \sigma_{\mathfrak{P}'} \tau^{-1} = \sigma_{\mathfrak{P}}$$

para algún $\tau \in G$.

Si el grupo de Galois $G = \operatorname{Gal}(L/K)$ es abeliano, entonces la conjugación es trivial, lo que implica que $\sigma_{\mathfrak{P}'} = \sigma_{\mathfrak{P}}$ para todos los primos sobre $\mathfrak{p}$. En este caso, el elemento de Frobenius depende únicamente del ideal primo $\mathfrak{p}$ de la base. Esto nos permite definir el **símbolo de Artin** (o símbolo de Frobenius):

$$\left(\frac{L/K}{\mathfrak{p}}\right) := \sigma_{\mathfrak{P}} \in \operatorname{Gal}(L/K)$$

El símbolo de Artin posee propiedades algebraicas y analíticas muy similares a las del símbolo de Legendre.

Si la extensión $L/K$ es abeliana, obtenemos un mapeo natural de los ideales primos no ramificados al grupo de Galois:

$$\mathfrak{p} \mapsto \left(\frac{L/K}{\mathfrak{p}}\right) \in \operatorname{Gal}(L/K)$$

¿Será posible que para cada automorfismo del grupo de Galois exista un primo asociado? La respuesta es afirmativa y está dada por el siguiente teorema fundamental:

> **Teorema de Densidad de Chebotarev (1922):** Para cualquier clase de conjugación $C \subseteq \operatorname{Gal}(L/K)$ (que en el caso abeliano consta de un único elemento $\sigma$), existen infinitos ideales primos no ramificados $\mathfrak{p}$ en $\mathcal{O}_K$ tales que:
>
> $$\left(\frac{L/K}{\mathfrak{p}}\right) = \sigma$$
>
> Además, estos primos están equidistribuidos con una densidad analítica (o de Dirichlet) igual a $|C|/|G|$.

Una consecuencia directa de este teorema es el **Teorema de Dirichlet sobre progresiones aritméticas**: al aplicar el Teorema de Chebotarev a extensiones ciclotómicas $L = \mathbb{Q}(\zeta_n)$ sobre $K = \mathbb{Q}$, se demuestra de inmediato que existen infinitos primos en cualquier progresión aritmética de la forma $a + kn$ con $\gcd(a, n) = 1$.


## Extensiones Ciclotómicas

Sea $\zeta_n$ una raíz $n$-ésima primitiva de la unidad. Se cumplen las siguientes propiedades fundamentales para la extensión ciclotómica $\mathbb{Q}(\zeta_n)/\mathbb{Q}$:

1. $[\mathbb{Q}(\zeta_n) : \mathbb{Q}] = \phi(n)$, donde $\phi$ es la función indicadora de Euler.
2. $\operatorname{Gal}(\mathbb{Q}(\zeta_n)/\mathbb{Q}) \cong (\mathbb{Z}/n\mathbb{Z})^\times$. Cualquier automorfismo de la extensión queda unívocamente determinado por saber adónde va la raíz de la unidad $\zeta_n$. Debe ser de la forma:
   $$\sigma_k(\zeta_n) = \zeta_n^k \quad \text{para algún } k \text{ con } \gcd(k, n) = 1$$

El automorfismo de Frobenius en este contexto es muy específico.

Si $p \nmid n$, entonces el primo $p$ no se ramifica en $\mathbb{Q}(\zeta_n)$ (ya que $p$ no divide al discriminante del cuerpo, el cual puede calcularse explícitamente como $\operatorname{disc}(\mathbb{Q}(\zeta_n)) = (-1)^{\phi(n)/2} n^{\phi(n)} \prod_{q \mid n} q^{-\frac{\phi(n)}{q-1}}$).

Como $p$ no se ramifica, tiene asociado un único elemento de Frobenius en el grupo de Galois. Este elemento, representado por el símbolo de Artin, actúa de la siguiente manera:

$$p \mapsto \left(\frac{\mathbb{Q}(\zeta_n)/\mathbb{Q}}{(p)}\right) = \sigma_p \quad \text{donde} \quad \sigma_p(\zeta_n) = \zeta_n^p$$

Recordemos que el orden de este automorfismo es exactamente el grado de inercia $f$ de $p$ en la extensión. Como $\sigma_p$ corresponde a la clase de $p$ en $(\mathbb{Z}/n\mathbb{Z})^\times$, el grado de inercia $f$ es igual al orden multiplicativo de $p$ módulo $n$.

Esto resulta sumamente interesante al analizar los dos casos extremos para la descomposición de un primo $p \nmid n$:

### Caso 1: Escisión completa (Splits completely)
Si $p \equiv 1 \pmod n$, entonces el orden multiplicativo de $p$ módulo $n$ es $\operatorname{ord}_n(p) = 1$. Por lo tanto, el grado de inercia es $f = 1$.
Como $e = 1$ (no ramificado), el número de factores primos es $g = \phi(n)/f = \phi(n)$. En este caso, el ideal primo $(p)$ se escinde completamente en la extensión ciclotómica:
$$p\mathcal{O}_L = \mathfrak{p}_1 \mathfrak{p}_2 \cdots \mathfrak{p}_{\phi(n)}$$

### Caso 2: Primos inertes (Inert primes)
Si $p$ es una raíz primitiva módulo $n$, entonces el orden de $p$ en $(\mathbb{Z}/n\mathbb{Z})^\times$ es máximo, es decir, $\operatorname{ord}_n(p) = \phi(n)$. 
Esto implica que el grado de inercia es $f = \phi(n)$ y, por tanto, el número de factores primos es $g = 1$. El ideal primo $(p)$ permanece primo en la extensión (es inerte):
$$p\mathcal{O}_L = \mathfrak{P}$$

Para que existan primos inertes en $\mathbb{Q}(\zeta_n)/\mathbb{Q}$, es necesario que el grupo de Galois $(\mathbb{Z}/n\mathbb{Z})^\times$ sea cíclico (es decir, que admita raíces primitivas). Por un resultado clásico de teoría de números, esto ocurre si y solo si:
$$n = 1, 2, 4, q^k, 2q^k \quad \text{donde } q \text{ es un primo impar.}$$

#### Ejemplo 1: El caso $n = 35$
Para el cuerpo ciclotómico $\mathbb{Q}(\zeta_{35})$, el grupo de Galois es:
$$\operatorname{Gal}(\mathbb{Q}(\zeta_{35})/\mathbb{Q}) \cong (\mathbb{Z}/35\mathbb{Z})^\times \cong (\mathbb{Z}/5\mathbb{Z})^\times \times (\mathbb{Z}/7\mathbb{Z})^\times \cong \mathbb{Z}/4\mathbb{Z} \times \mathbb{Z}/6\mathbb{Z}$$
Este grupo no es cíclico. El orden máximo de cualquier elemento en $\mathbb{Z}/4\mathbb{Z} \times \mathbb{Z}/6\mathbb{Z}$ es $\operatorname{mcm}(4, 6) = 12$, el cual es menor que $\phi(35) = 24$.
Como ningún primo $p$ puede tener orden multiplicativo $24$ módulo $35$, el grado de inercia $f$ para cualquier primo no ramificado $p$ es a lo sumo $12$. Esto significa que $g = 24/f \ge 2$. Por lo tanto, **no existen primos inertes en $\mathbb{Q}(\zeta_{35})/\mathbb{Q}$**.

#### Ejemplo 2: El caso $n = 8$
Para $\mathbb{Q}(\zeta_{8})$, tenemos $\phi(8) = 4$. El grupo de Galois es:
$$\operatorname{Gal}(\mathbb{Q}(\zeta_{8})/\mathbb{Q}) \cong (\mathbb{Z}/8\mathbb{Z})^\times \cong \mathbb{Z}/2\mathbb{Z} \times \mathbb{Z}/2\mathbb{Z}$$
El orden de cualquier elemento en este grupo (distinto de la identidad) es $2$. Así, para cualquier primo impar $p$, el grado de inercia es $f \le 2$, de modo que $g = 4/f \ge 2$. Por lo tanto, **tampoco existen primos inertes en $\mathbb{Q}(\zeta_8)/\mathbb{Q}$**.

#### Ejemplo 3: El caso $n = 15$
Para $\mathbb{Q}(\zeta_{15})$, tenemos $\phi(15) = 8$. El grupo de Galois es:
$$\operatorname{Gal}(\mathbb{Q}(\zeta_{15})/\mathbb{Q}) \cong (\mathbb{Z}/15\mathbb{Z})^\times \cong (\mathbb{Z}/3\mathbb{Z})^\times \times (\mathbb{Z}/5\mathbb{Z})^\times \cong \mathbb{Z}/2\mathbb{Z} \times \mathbb{Z}/4\mathbb{Z}$$
El orden máximo de cualquier elemento es $\operatorname{mcm}(2, 4) = 4 < 8$. En consecuencia, el grado de inercia de cualquier primo impar es a lo sumo $4$, lo que exige que $g = 8/f \ge 2$. Por tanto, **no hay primos inertes en $\mathbb{Q}(\zeta_{15})/\mathbb{Q}$**.

### Ejemplo Detallado: El caso $n = 5$ ($\mathbb{Q}(\zeta_5)/\mathbb{Q}$)

Para $n = 5$, la extensión ciclotómica tiene grado $\phi(5) = 4$. El grupo de Galois es $\operatorname{Gal}(\mathbb{Q}(\zeta_5)/\mathbb{Q}) \cong (\mathbb{Z}/5\mathbb{Z})^\times \cong \mathbb{Z}/4\mathbb{Z}$, el cual es cíclico. 
Analicemos el comportamiento de un primo $p \neq 5$:

1. **Si $p \equiv 1 \pmod 5$:**
   * El orden es $\operatorname{ord}_5(p) = 1 \implies f = 1$ y $g = 4$.
   * El ideal $(p)$ se escinde completamente en cuatro ideales primos distintos.
   * **Ejemplos:** $11, 31, 41, 61, 71$.
   
2. **Si $p \equiv 2 \pmod 5$ o $p \equiv 3 \pmod 5$:**
   * Las clases de $2$ y $3$ son raíces primitivas módulo $5$, por lo que $\operatorname{ord}_5(p) = 4 \implies f = 4$ y $g = 1$.
   * El ideal $(p)$ es **inerte** (se mantiene como un único ideal primo).
   * **Ejemplos:** $2, 3, 7, 13, 17, 23, 37, 43, 47, 53, 73, 83$.
   * Como $103 \equiv 3 \pmod 5$, entonces el ideal $103\mathcal{O}_{\mathbb{Q}(\zeta_5)}$ es inerte.
   
3. **Si $p \equiv 4 \pmod 5$:**
   * Como $4 \equiv -1 \pmod 5$, el orden es $\operatorname{ord}_5(p) = 2 \implies f = 2$ y $g = 2$.
   * El ideal $(p)$ se descompone en el producto de dos ideales primos distintos, cada uno con grado de inercia $2$:
     $$p\mathcal{O}_{\mathbb{Q}(\zeta_5)} = \mathfrak{p}\mathfrak{q}$$
   * **Ejemplos:** $19, 29, 59, 79, 89$.

### Ejemplo Adicional: El caso $n = 7$ ($\mathbb{Q}(\zeta_7)/\mathbb{Q}$)

Para $n = 7$, la extensión tiene grado $\phi(7) = 6$. El grupo de Galois es $\operatorname{Gal}(\mathbb{Q}(\zeta_7)/\mathbb{Q}) \cong (\mathbb{Z}/7\mathbb{Z})^\times \cong \mathbb{Z}/6\mathbb{Z}$, también cíclico.
Para un primo $p \neq 7$:

1. **Si $p \equiv 1 \pmod 7$:**
   * $\operatorname{ord}_7(p) = 1 \implies f = 1$ y $g = 6$ (se escinde completamente en $6$ ideales).
   * **Ejemplos:** $29, 43, 71, 113$.

2. **Si $p \equiv 3 \pmod 7$ o $p \equiv 5 \pmod 7$:**
   * Las clases de $3$ y $5$ son raíces primitivas módulo $7$, por lo que $\operatorname{ord}_7(p) = 6 \implies f = 6$ y $g = 1$.
   * El ideal $(p)$ es **inerte**.
   * **Ejemplos:** $3, 5, 17, 19, 31, 47, 59, 61, 89$.

3. **Si $p \equiv 2 \pmod 7$ o $p \equiv 4 \pmod 7$:**
   * Dado que $2^3 \equiv 1 \pmod 7$ y $4^3 \equiv 1 \pmod 7$, tenemos $\operatorname{ord}_7(p) = 3 \implies f = 3$ y $g = 2$.
   * El ideal $(p)$ se descompone como el producto de dos ideales primos, cada uno con grado de inercia $3$:
     $$p\mathcal{O}_{\mathbb{Q}(\zeta_7)} = \mathfrak{p}\mathfrak{q}$$
   * **Ejemplos:** $2, 11, 23, 37, 53, 67, 79$.

4. **Si $p \equiv 6 \pmod 7$:**
   * Como $6 \equiv -1 \pmod 7$, el orden es $\operatorname{ord}_7(p) = 2 \implies f = 2$ y $g = 3$.
   * El ideal $(p)$ se descompone en el producto de tres ideales primos con grado de inercia $2$:
     $$p\mathcal{O}_{\mathbb{Q}(\zeta_7)} = \mathfrak{p}_1 \mathfrak{p}_2 \mathfrak{p}_3$$
   * **Ejemplos:** $13, 41, 83, 97$.

## El Mapeo de Artin y los Grupos de Clases de Rayos

Hasta ahora hemos definido el símbolo de Artin para ideales primos no ramificados. Si la extensión $L/K$ es abeliana, podemos extender esta definición de manera multiplicativa a cualquier ideal fraccionario.

Sea $S$ el conjunto de ideales primos de $K$ que se ramifican en $L$. Denotamos por $I_K(S)$ al grupo de ideales fraccionarios de $K$ generados por los primos que no pertenecen a $S$. Definimos el **mapeo de Artin** (o aplicación de Artin) como el homomorfismo:

$$\Phi_{L/K}: I_K(S) \to \operatorname{Gal}(L/K)$$
$$\mathfrak{a} = \prod_{i} \mathfrak{p}_i^{a_i} \mapsto \prod_{i} \left(\frac{L/K}{\mathfrak{p}_i}\right)^{a_i}$$

El gran avance de la Teoría de Cuerpos de Clase consiste en describir el kernel de este mapeo y relacionarlo con subgrupos de ideales definidos únicamente en términos del cuerpo base $K$. Para ello, introducimos los conceptos de **módulo** y **grupo de clases de rayos**.

### Módulos y el Grupo de Rayos

1. **Módulo ($\mathfrak{m}$):**
   Un módulo en $K$ es un producto formal $\mathfrak{m} = \mathfrak{m}_0 \mathfrak{m}_\infty$, donde:
   * $\mathfrak{m}_0$ es un ideal entero de $\mathcal{O}_K$.
   * $\mathfrak{m}_\infty$ es un conjunto (posiblemente vacío) de inmersiones reales $\sigma: K \hookrightarrow \mathbb{R}$.

2. **Grupo de ideales coprimos ($I_K(\mathfrak{m})$):**
   Es el subgrupo de ideales fraccionarios de $K$ generados por primos que no dividen a $\mathfrak{m}_0$.

3. **Subgrupo de Rayo Principal ($P_{K,1}(\mathfrak{m})$):**
   Es el subgrupo de ideales fraccionarios principales generados por elementos $\alpha \in K^\times$ tales que:
   * $\alpha \equiv 1 \pmod{\mathfrak{m}_0}$ (es decir, $v_{\mathfrak{p}}(\alpha - 1) \ge v_{\mathfrak{p}}(\mathfrak{m}_0)$ para todo $\mathfrak{p} \mid \mathfrak{m}_0$).
   * $\sigma(\alpha) > 0$ para toda inmersión real $\sigma \in \mathfrak{m}_\infty$.

4. **Grupo de Clases de Rayos (Ray Class Group, $C_K(\mathfrak{m})$):**
   Se define como el cociente:
   $$C_K(\mathfrak{m}) = I_K(\mathfrak{m}) / P_{K,1}(\mathfrak{m})$$
   Este grupo es siempre finito y generaliza al grupo de clases de ideales clásico $C_K$ (el cual se recupera cuando $\mathfrak{m} = 1$).

### El Teorema de Reciprocidad de Artin

El teorema central de la teoría de cuerpos de clase global (demostrado por Emil Artin en 1927) establece lo siguiente:

> **Teorema de Reciprocidad de Artin:** Sea $L/K$ una extensión abeliana finita. Existe un módulo $\mathfrak{m}$ de $K$ (que contiene a todos los primos ramificados de $K$ en $L$) tal que el mapeo de Artin $\Phi_{L/K}: I_K(\mathfrak{m}) \to \operatorname{Gal}(L/K)$ es sobreyectivo y su kernel contiene a $P_{K,1}(\mathfrak{m})$.

Como consecuencia, el mapeo de Artin induce un isomorfismo:
$$I_K(\mathfrak{m}) / H \cong \operatorname{Gal}(L/K)$$
donde $H$ (el kernel de $\Phi_{L/K}$) es un subgrupo de congruencia que satisface $P_{K,1}(\mathfrak{m}) \subseteq H \subseteq I_K(\mathfrak{m})$.

El menor módulo $\mathfrak{f} = \mathfrak{f}(L/K)$ para el cual se cumple este teorema se conoce como el **conductor** de la extensión. Un primo $\mathfrak{p}$ de $K$ se ramifica en $L$ si y solo si divide al conductor $\mathfrak{f}$.


## El Cuerpo de Clases de Hilbert y la ecuación $p = x^2 + 5y^2$

Sea $K$ un cuerpo de números. Definimos el **cuerpo de clases de Hilbert** de $K$, denotado por $H_K$, como la extensión abeliana finita más grande de $K$ que es no ramificada en todos los primos (tanto finitos como infinitos).

Un teorema fundamental en la Teoría de Cuerpos de Clase establece la relación directa con el grupo de clases:

> **Teorema del Cuerpo de Clases de Hilbert:** Si $H_K$ es el cuerpo de clases de Hilbert de $K$, entonces el mapeo de Artin induce un isomorfismo:
> $$\operatorname{Cl}(K) \cong \operatorname{Gal}(H_K/K)$$
> donde $\operatorname{Cl}(K)$ es el grupo de clases de ideales clásico de $K$. Este isomorfismo asocia cada clase de ideales $[\mathfrak{p}]$ con el automorfismo de Frobenius $\left(\frac{H_K/K}{\mathfrak{p}}\right)$.

El grupo de clases $\operatorname{Cl}(K)$ mide qué tan diferente de un Dominio de Ideales Principales (PID) es el anillo $\mathcal{O}_K$. Por este teorema, existe una extensión abeliana no ramificada de $K$ cuyo grupo de Galois es exactamente el grupo de clases.

### Aplicación: La ecuación $p = x^2 + 5y^2$

Las soluciones de la ecuación diofántica $p = x^2 + 5y^2$ provienen de la factorización en el anillo de enteros $\mathcal{O}_K = \mathbb{Z}[\sqrt{-5}]$ del cuerpo cuadrático imaginario $K = \mathbb{Q}(\sqrt{-5})$.

Como sabemos, $\mathbb{Z}[\sqrt{-5}]$ no es un PID. Su grupo de clases es:
$$\operatorname{Cl}(K) \cong \mathbb{Z}/2\mathbb{Z}$$
Esto significa que todos los ideales de $\mathcal{O}_K$ son principales, o bien sus cuadrados son principales.

Por el teorema de Hilbert, el cuerpo de clases de Hilbert $H_K$ debe ser una extensión cuadrática de $K$ (ya que $[H_K:K] = |\operatorname{Cl}(K)| = 2$). Se puede demostrar que:
$$H_K = \mathbb{Q}(\sqrt{-5}, i)$$

Para verificar que este es efectivamente el cuerpo de clases de Hilbert, basta comprobar que $H_K/K$ es una extensión no ramificada. Su discriminante relativo es el ideal trivial $d_{H_K/K} = (1)$, lo que confirma que ningún primo se ramifica en esta extensión.

Por lo tanto, si un primo $p \in \mathbb{Z}$ se puede escribir como $p = x^2 + 5y^2$, entonces en $K = \mathbb{Q}(\sqrt{-5})$ se factoriza como $p = a\bar{a}$, con $a = x + \sqrt{-5}y \in \mathcal{O}_K$. Esto implica que:
1. El primo $p$ tiene que descomponerse en la extensión $K/\mathbb{Q}$.
2. Sus factores primos en $\mathcal{O}_K$ deben ser **ideales principales**.


### La Conexión Final: El Diagrama de Cuerpos

La existencia de una solución a $p = x^2 + 5y^2$ es equivalente a que el primo $p$ se descomponga en factores primos principales (como ideales) en $K = \mathbb{Q}(\sqrt{-5})$. Por la teoría de cuerpos de clase de Hilbert, un ideal primo $\mathfrak{p}$ en $K$ es principal si y solo si se descompone completamente en el cuerpo de clases de Hilbert $H_K = \mathbb{Q}(\sqrt{-5}, i)$. 

Por lo tanto, la ecuación tiene solución si y solo si el primo $p$ se descompone completamente en la extensión bicuadrática $H_K / \mathbb{Q}$.

Consideremos el siguiente diagrama de subcuerpos:

```
             Q(√-5, i)
            /    |    \
           /     |     \
      Q(√-5)    Q(i)   Q(√5)
           \     |     /
            \    |    /
                 Q
```

Dado que el grupo de Galois $\operatorname{Gal}(H_K/\mathbb{Q})$ es el grupo de Klein $V_4 \cong \mathbb{Z}/2\mathbb{Z} \times \mathbb{Z}/2\mathbb{Z}$, un primo $p$ se descompone completamente en $H_K$ si y solo si se descompone completamente en al menos dos (y por ende en los tres) de sus subcuerpos cuadráticos.

Así, para un primo impar $p \neq 5$, $p$ se puede expresar de la forma $x^2 + 5y^2$ si y solo si se descompone en:
1. **$\mathbb{Q}(i)$:** Esto ocurre si y solo si $p \equiv 1 \pmod 4$.
2. **$\mathbb{Q}(\sqrt{5})$:** Esto ocurre si y solo si $\left(\frac{5}{p}\right) = 1 \iff p \equiv \pm 1 \pmod 5$.

Al combinar estas condiciones mod $4$ y mod $5$ usando el **Teorema Chino del Residuo**, obtenemos:
* $\begin{cases} p \equiv 1 \pmod 4 \\ p \equiv 1 \pmod 5 \end{cases} \implies p \equiv 1 \pmod{20}$
* $\begin{cases} p \equiv 1 \pmod 4 \\ p \equiv 4 \pmod 5 \end{cases} \implies p \equiv 9 \pmod{20}$

> **Conclusión:** Un primo impar $p \neq 5$ se puede escribir de la forma $x^2 + 5y^2$ si y solo si:
> $$p \equiv 1 \quad \text{o} \quad 9 \pmod{20}$$
> (Adicionalmente, el primo ramificado $p = 5$ también tiene la solución trivial $5 = 0^2 + 5 \cdot 1^2$).

### El Teorema del Ideal Principal y las Torres de Cuerpos de Clase

Finalmente, concluimos esta sesión con el siguiente teorema fundamental de la teoría de cuerpos de clase:

> **Teorema del Ideal Principal:** Si $H_K$ es el cuerpo de clases de Hilbert de $K$, entonces todo ideal de $\mathcal{O}_K$ se vuelve principal al extenderse a $\mathcal{O}_{H_K}$.

#### Ejemplo
En $\mathcal{O}_K = \mathbb{Z}[\sqrt{-5}]$, el ideal $\mathfrak{a} = (2, 1 + \sqrt{-5})$ no es principal. Sin embargo, al extenderlo al cuerpo de clases de Hilbert $H_K = \mathbb{Q}(\sqrt{-5}, i)$, se vuelve principal y está generado por el elemento $1+i$:
$$\mathfrak{a}\mathcal{O}_{H_K} = (1+i)\mathcal{O}_{H_K}$$

A partir de este resultado, surge una pregunta natural: si denotamos $K_1 = H_K$, todos los ideales de $K$ son principales en $K_1$, pero ¿son todos los ideales de $\mathcal{O}_{K_1}$ principales? No necesariamente. Esto nos permite definir una secuencia de extensiones sucesivas conocida como la **torre de cuerpos de clase**:
$$K \to K_1 \to K_2 \to K_3 \to \dots$$
donde cada $K_{i+1}$ es el cuerpo de clases de Hilbert del cuerpo precedente $K_i$.

El estudio de si esta sucesión es siempre finita (es decir, si existe un paso a partir del cual todo ideal es principal) o si puede ser infinita fue uno de los problemas más importantes de la teoría de números algebraica. En 1964, Golod y Shafarevich demostraron que existen torres de cuerpos de clase infinitas. 

En particular, el cuerpo cuadrático imaginario $\mathbb{Q}(\sqrt{-5460})$ es el más pequeño para el cual **aún no se sabe** si su torre de cuerpos de clase termina o no (un problema abierto estrechamente vinculado con la teoría de generadores de Galois).
